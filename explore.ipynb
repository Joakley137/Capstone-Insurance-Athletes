```python
# Any SQL works; :named params go in a dict
sdb.query("SELECT body_part, COUNT(*) AS n, ROUND(AVG(days_missed), 1) AS avg_days FROM injuries "
          "WHERE sport = :sport AND body_part IS NOT NULL GROUP BY body_part ORDER BY n DESC LIMIT 10",
          {"sport": "Soccer"})
```
