# Explore the sports_injury database

Connects to MySQL through `sportsdb.py` and pulls each table into pandas.
MySQL must be running (`brew services start mysql`).

```python
import sportsdb as sdb

sdb.tables()   # every table and view, with row counts
```

```python
sdb.injuries(sport="NFL").head()
```

```python
sdb.contracts(sport="NFL", min_apy=30_000_000)
```

```python
sdb.contracts_with_injuries(sport="NFL").head()   # contracts + career injury totals
```

```python
sdb.body_part_rates(sport="Soccer")
```

```python
sdb.insurance_cases()
```

```python
# Any SQL works; :named parameters are passed as a dict
sdb.query("SELECT team, COUNT(*) AS n FROM injuries WHERE sport = :sport GROUP BY team ORDER BY n DESC LIMIT 10",
          {"sport": "NFL"})
```
